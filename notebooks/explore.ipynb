## Step 1 – Load both years (all airports)

`load_year()` reads a file, drops summary/subtotal rows (`Rank = NaN`), and **detects** the current-year enplanement column automatically — it prints the detected name so you can verify it.

No `S/L` filtering here. We keep every airport from both files so the 2019 lookup in Step 2 can match airports regardless of their 2019 classification.

```python
import warnings

KEEP_COLS = ["Locid", "ST", "City", "Airport Name", "S/L", "Hub"]

def load_year(year: int) -> pd.DataFrame:
    path = f"data/raw/arp-cy{year}-commercial-service-enplanements.xlsx"
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        raw = pd.read_excel(path)
    raw = raw[raw["Rank"].notna()].copy()
    enpl_col = next(c for c in raw.columns if "Enplanements" in c and str(year)[-2:] in c)
    print(f"  {year}: detected enplanement column → '{enpl_col}'")
    raw = raw[KEEP_COLS + [enpl_col]].rename(columns={enpl_col: "enplanements"})
    raw["year"] = year
    return raw

df_2019 = load_year(2019)
df_2025 = load_year(2025)

long = pd.concat([df_2019, df_2025], ignore_index=True)
long = long.rename(columns={"Locid": "code", "ST": "state", "City": "city",
                             "Airport Name": "name", "S/L": "service_level", "Hub": "hub"})

CODE_ALIASES = {"PBI": "DJT"}    # old code -> current code
long["code"] = long["code"].replace(CODE_ALIASES)

print(f"\nLong table shape: {long.shape}  (all airports, both years)")
long.head(4)
```

## Step 2 – Eligibility: Primary in 2025

Eligibility is determined **solely by 2025 status** (`service_level == "P"`). That gives us the universe of airports we want to score.

For those airports we then look up their 2019 enplanements from the 2019 file — **regardless of what their 2019 `S/L` was**. An airport that was `CS` in 2019 but `P` in 2025 still gets its 2019 passenger count included; its growth is real and we want to capture it.

```python
# Eligible airports = Primary in 2025
eligible_codes = set(long.loc[(long["year"] == 2025) & (long["service_level"] == "P"), "code"])
print(f"Primary airports in 2025: {len(eligible_codes)}")

# Base table: 2025 rows for eligible airports (carries hub, state, name for peer groups)
base = long[(long["year"] == 2025) & (long["code"].isin(eligible_codes))].copy()
base = base.rename(columns={"enplanements": "enpl_2025"})

# 2019 enplanements for those airports, regardless of their 2019 S/L
enpl_2019 = (
    long.loc[(long["year"] == 2019) & (long["code"].isin(eligible_codes)),
             ["code", "enplanements"]]
    .rename(columns={"enplanements": "enpl_2019"})
)
print(f"Of those, with 2019 data:  {len(enpl_2019)}")
```

## Step 3 – Compute 2019 → 2025 CAGR (with growth_flag)

**CAGR formula:** `(enpl_2025 / enpl_2019) ^ (1 / 6) - 1`

We **left-join** 2019 data onto the 2025 Primary airports — no rows are dropped. Airports without a 2019 match get `cagr_19_25 = NaN` and `growth_flag = "insufficient_history"` (per D5). All others get `growth_flag = "computed"`.

```python
N_YEARS = 6

# Left join: keep all 2025 Primary airports, attach 2019 where available
cagr_df = base.merge(enpl_2019, on="code", how="left")

has_2019 = cagr_df["enpl_2019"].notna()
cagr_df["cagr_19_25"] = (cagr_df["enpl_2025"] / cagr_df["enpl_2019"]) ** (1 / N_YEARS) - 1
cagr_df["growth_flag"] = "computed"
cagr_df.loc[~has_2019, "growth_flag"] = "insufficient_history"

print(f"Primary airports in 2025:                    {len(cagr_df)}")
print(f"  with 2019 data  (growth_flag=computed):    {has_2019.sum()}")
print(f"  missing 2019    (growth_flag=insufficient): {(~has_2019).sum()}")

cagr_df = cagr_df.sort_values("cagr_19_25", ascending=False).reset_index(drop=True)
cagr_df[["code", "city", "state", "hub", "enpl_2019", "enpl_2025", "cagr_19_25", "growth_flag"]].head(10)
```

## T-100 vs FAA enplanements spot-check

T-100 segment data (BTS) sums passengers by origin airport. FAA enplanements count every boarding. They should be close — a gap of a few percent is normal. A large gap flags a data problem.

**What to expect:**
- Domestic-focused airports → close match (within ~3%)
- International hubs → T-100 will be systematically *lower*, because `T_T100D` is domestic-only; their international legs are in a separate `T_T100I` file
- Alaska airports → T-100 can be *higher* due to multi-stop "milk run" routing where T-100 counts each segment separately

## Build airport_metrics

One row per FAA Primary airport (2025). Combines:
- `cagr_df` — base: FAA code, hub, enplanements 2019/2025, CAGR, data_confidence
- `faa` — enpl_2024 and pct_change_24_25
- T-100 segment metrics — capacity, load factor, long-haul share, cargo share (D13/D14)
- BTS delay metrics — delay rate, NAS delay rate, cancel rate (D12)

Join key: FAA code → IATA via `faa_to_iata` (OurAirports + manual overrides, D9).

```python
# ── T-100 airport metrics (D13: origin-side departures, D14: per-metric CLASS) ─
# Rule: aggregate sums first, then compute ratios — never average rates

fl_rows  = t[t["CLASS"].isin(["F", "L"])]            # scheduled + charter pax
f_rows   = t[t["CLASS"] == "F"]                       # scheduled pax only
lh_f     = f_rows[f_rows["DISTANCE"] >= LONG_HAUL_MILES]
lh_all   = t[t["DISTANCE"] >= LONG_HAUL_MILES]
cargo    = t[t["CLASS"].isin(["G", "P"])]              # all-cargo

fl_agg   = fl_rows.groupby("ORIGIN")[["DEPARTURES_PERFORMED", "PASSENGERS"]].sum()
fl_agg.columns = ["departures_FL", "passengers_FL"]

f_agg    = f_rows.groupby("ORIGIN")[["DEPARTURES_PERFORMED", "SEATS", "PASSENGERS"]].sum()
f_agg.columns   = ["_dep_F", "_seats_F", "_pax_F"]

t100_m = (
    fl_agg
    .join(f_agg, how="outer")
    .join(t.groupby("ORIGIN")["DEPARTURES_PERFORMED"].sum().rename("tot_departures_all"), how="outer")
    .join(cargo.groupby("ORIGIN")["DEPARTURES_PERFORMED"].sum().rename("_cargo_dep"), how="outer")
    .join(lh_f.groupby("ORIGIN")["DEPARTURES_PERFORMED"].sum().rename("_lh_dep_F"), how="outer")
    .join(lh_f.groupby("ORIGIN")["PASSENGERS"].sum().rename("_lh_pax_F"), how="outer")
    .join(lh_all.groupby("ORIGIN")["DEPARTURES_PERFORMED"].sum().rename("_lh_dep_all"), how="outer")
    .reset_index()
    .rename(columns={"ORIGIN": "iata"})
)

t100_m["load_factor"]                = t100_m["_pax_F"]      / t100_m["_seats_F"]
t100_m["long_haul_share_flights"]    = t100_m["_lh_dep_F"]   / t100_m["_dep_F"]
t100_m["long_haul_share_pax"]        = t100_m["_lh_pax_F"]   / t100_m["_pax_F"]
t100_m["long_haul_share_all_flights"]= t100_m["_lh_dep_all"] / t100_m["tot_departures_all"]
t100_m["cargo_departure_share"]      = t100_m["_cargo_dep"]  / t100_m["tot_departures_all"]

t100_m = t100_m.drop(columns=["_dep_F", "_seats_F", "_pax_F",
                                "_cargo_dep", "_lh_dep_F", "_lh_pax_F", "_lh_dep_all"])

print(f"T-100 metrics: {t100_m.shape}, {t100_m['iata'].nunique()} unique airports")
t100_m.head(3)
```

```python
# ── Delay metrics per airport (D12: sum carriers + months first, then divide) ─
delay_sums = (
    delays
    .groupby("airport")[["arr_flights", "arr_del15", "nas_ct", "arr_cancelled"]]
    .sum()
)
delay_m = delay_sums.copy()
delay_m["delay_rate"]     = delay_m["arr_del15"]    / delay_m["arr_flights"]
delay_m["nas_delay_rate"] = delay_m["nas_ct"]        / delay_m["arr_flights"]
delay_m["cancel_rate"]    = delay_m["arr_cancelled"] / delay_m["arr_flights"]

delay_m = (
    delay_m[["arr_flights", "delay_rate", "nas_delay_rate", "cancel_rate"]]
    .reset_index()
    .rename(columns={"airport": "iata"})
)
print(f"Delay metrics: {delay_m.shape}, {delay_m['iata'].nunique()} unique airports")
```

```python
# ── Assemble airport_metrics: one row per Primary airport (2025) ──────────────
am = cagr_df[["code", "state", "city", "name", "hub",
               "enpl_2025", "enpl_2019", "cagr_19_25", "growth_flag"]].copy()

# IATA mapping with T-100 fallback (D9):
#   1. use mapped IATA if it appears in T-100 ORIGIN
#   2. else if FAA code itself is in T-100 ORIGIN, use the FAA code directly
#   3. else use mapped IATA as best effort
t100_origins = set(t100_m["iata"])

def _map_iata(faa_code):
    mapped = faa_to_iata.get(faa_code, faa_code)
    if mapped in t100_origins:
        return mapped
    if faa_code in t100_origins:
        return faa_code
    return mapped

am["iata"] = am["code"].map(_map_iata)

# Regression: any airport whose FAA code matched T-100 directly should not lose
# its match after the mapping — this list must be empty
direct_match = am["code"].isin(t100_origins)
reg_fail = am.loc[direct_match & ~am["iata"].isin(t100_origins), ["code", "iata", "city"]]
print(f"Regression (expect 0): FAA→T-100 direct match lost by mapping: {len(reg_fail)}")
if not reg_fail.empty:
    print(reg_fail)

# enpl_2024 and pct_change_24_25 from FAA table
am = am.merge(faa[["code", "enpl_2024", "pct_change_24_25"]], on="code", how="left")

# LEFT JOIN T-100 metrics (keyed by IATA)
am = am.merge(t100_m, on="iata", how="left")
am["has_t100"] = am["iata"].isin(t100_origins)

# Cross-validation: T-100 passengers (CLASS F+L, departures side, IATA-mapped)
# vs FAA enplanements — stored as a fraction; NaN when airport has T-100 data
# but no CLASS F+L service (e.g. pure charter/cargo destination)
am["t100_vs_faa_diff"] = np.where(
    am["has_t100"],
    (am["passengers_FL"] - am["enpl_2025"]) / am["enpl_2025"],
    np.nan
)

# Data confidence based on IATA-mapped cross-validation:
# "no_t100" only when has_t100 is False (1 airport: Ceiba PR)
# "high"/"low" threshold = 10% absolute gap; NaN diff (no F+L service) → "low"
am["data_confidence"] = "no_t100"
am.loc[am["has_t100"] & (am["t100_vs_faa_diff"].abs() <= 0.10), "data_confidence"] = "high"
am.loc[am["has_t100"] & (am["t100_vs_faa_diff"].abs() > 0.10), "data_confidence"] = "low"
am.loc[am["has_t100"] & am["t100_vs_faa_diff"].isna(), "data_confidence"] = "low"

# fillna(0) for share cols — only for airports present in T-100
# (0 = no long-haul routes; NaN = airport not in T-100 at all)
_share_cols = ["long_haul_share_flights", "long_haul_share_pax",
               "long_haul_share_all_flights", "cargo_departure_share"]
for col in _share_cols:
    am.loc[am["has_t100"] & am[col].isna(), col] = 0.0

# LEFT JOIN delay metrics (keyed by IATA)
am = am.merge(delay_m, on="iata", how="left")
am["has_delays"] = am["iata"].isin(set(delay_m["iata"]))

# ── Checks ────────────────────────────────────────────────────────────────────
print(f"\nShape: {am.shape}  (expect 402 rows)")
print(f"Unique codes: {am['code'].nunique()}")
print()
print("has_t100 (expect True=401, False=1 [RVR]):")
print(am["has_t100"].value_counts())
print()
print("has_delays:")
print(am["has_delays"].value_counts())
print()
print("Airports missing T-100 (has_t100 == False):")
print(am.loc[~am["has_t100"], ["code", "iata", "city", "state", "hub", "enpl_2025"]].to_string(index=False))
print()
print("data_confidence (expect no_t100=1):")
print(am["data_confidence"].value_counts())
print()
print("Code-mapped airports — IWA, GPI, DJT, GSN, 1G4:")
print(am[am["code"].isin(["IWA", "GPI", "DJT", "GSN", "1G4"])][
    ["code", "iata", "enpl_2025", "passengers_FL", "t100_vs_faa_diff", "data_confidence"]
].to_string(index=False))
print()
print("Spot check BOS / LAX / SNA / SFO / ANC:")
am[am["code"].isin(["BOS", "LAX", "SNA", "SFO", "ANC"])].set_index("code").T
```
